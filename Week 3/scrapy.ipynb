- Inverse Binary Thresholding

- Truncate Thresholding
    - th, dst_trunc = cv2.threshold(srcImg, thresholdVal, maxVal, cv2.THRESH_TRUNC)
- Threshold to zero
    - th, dst_to_zero = cv2.threshold(srcImg, thresholdVal, maxVal, cv2.THRESH_TOZERO)
- Inverted Threshold to zero
    - th, dst_to_zero_inv = cv2.threshold(srcImg, thresholdVal, maxVal, cv2.THRESH_TOZERO_INV)

```python
img = cv2.imread("data/images/threshold.png", cv2.IMREAD_GRAYSCALE)
thresh = 100
maxVal = 255
img.shape
```

```python
def thresholdingUsingLoop(img, threshVal, maxVal):
    height, width = img.shape[:2]
    dst = img.copy() # Output image

    # loop over rows
    for i in range(height):
        # loop over columns
        for j in range(width):
            if img[i, j] > threshVal:
                dst[i, j] = maxVal
            else:
                dst[i, j] = 0

    return dst

def thresholdUsingVectors(img, threshVal, maxVal):
    dst = np.zeros_like(img)
    dst[img>threshVal] = maxVal

    return dst

import time

t = time.time()
dst = thresholdingUsingLoop(img, thresh, maxVal)
print(f"Time taken = {time.time() - t} seconds.")

plt.figure(figsize=[15, 15])
plt.subplot(121);plt.imshow(img, cmap="gray");plt.title("Original Image")
plt.subplot(122);plt.imshow(dst, cmap="gray");plt.title("Thresholded Image")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

im = np.zeros((10, 10), dtype="uint8")

im[0,1], im[-1,0], im[-2,-1], im[2,2], im[5:8, 5:8] = 1,1,1,1,1

kernel1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
ksize = kernel1.shape[0]
height, width = im.shape[:2]
border = ksize//2

paddedIm = cv2.copyMakeBorder(im, border, border, border, border, cv2.BORDER_CONSTANT, value=0)
dilated = np.zeros(im.shape)

# Store frames for later display
frames = []

# Add original image
frames.append(im.copy())

change_count = 0

for hi in range(border, height+border):
    for wi in range(border, width+border): 
        found_white = False
        
        for i in range(ksize):
            for j in range(ksize):
                if kernel1[i, j] == 1:
                    img_r = hi + (i - border)
                    img_c = wi + (j - border)
                    
                    if paddedIm[img_r, img_c] == 1:
                        found_white = True
                        break
            if found_white:
                break
        
        output_row = hi - border
        output_col = wi - border
        
        old_value = dilated[output_row, output_col]
        dilated[output_row, output_col] = 1 if found_white else 0
        
        # Capture frame when pixel changes
        if old_value == 0 and dilated[output_row, output_col] == 1:
            change_count += 1
            frames.append(dilated.copy())

# Add final result
frames.append(dilated.copy())
```

```python
plt.figure(figsize=[10,10])
plt.subplot(121);plt.imshow(frames[0], cmap='gray');plt.title("Original Image")
plt.subplot(122);plt.imshow(frames[-1], cmap='gray');plt.title("Final Image")
```

```python
fps = 30
frame_width, frame_height = frames[0].shape[:2]
out = cv2.VideoWriter('dilation_process.mp4', cv2.VideoWriter_fourcc('M','J','P','G'), fps, (frame_width, frame_height), isColor=False)

print(f"\nWriting video...")
for idx, frame in enumerate(frames):
    out.write(frame)

out.release()
print("Done writing into a .mp4 file")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

im = np.zeros((10, 10), dtype="uint8")
im[0,1], im[-1,0], im[-2,-1], im[2,2], im[5:8, 5:8] = 1,1,1,1,1

kernel1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3,3))
ksize = kernel1.shape[0]
height, width = im.shape[:2]
border = ksize//2

paddedIm = cv2.copyMakeBorder(im, border, border, border, border, cv2.BORDER_CONSTANT, value=0)
eroded = np.zeros(im.shape)

# Store frames
frames = []

# Add original image
frames.append(im.copy())

change_count = 0

for hi in range(border, height+border):
    for wi in range(border, width+border): 
        
        all_white = True
        
        for i in range(ksize):
            for j in range(ksize):
                if kernel1[i, j] == 1:
                    img_r = hi + (i - border)
                    img_c = wi + (j - border)
                    
                    if paddedIm[img_r, img_c] == 0:
                        all_white = False
                        break
            
            if not all_white:
                break
        
        output_row = hi - border
        output_col = wi - border
        
        old_value = eroded[output_row, output_col]
        eroded[output_row, output_col] = 1 if all_white else 0
        
        # Capture frame when pixel changes
        if old_value != eroded[output_row, output_col]:
            change_count += 1
            frames.append(eroded.copy())

# Add final result
frames.append(eroded.copy())
```

```python
plt.figure(figsize=[10,10])
plt.subplot(121);plt.imshow(frames[0], cmap='gray');plt.title("Original Image")
plt.subplot(122);plt.imshow(frames[-1], cmap='gray');plt.title("Final Image")
```

```python
fps = 30
frame_width, frame_height = frames[0].shape[:2]
out = cv2.VideoWriter('erosion_process.mp4', cv2.VideoWriter_fourcc('M','J','P','G'), fps, (frame_width, frame_height), isColor=False)

print(f"\nWriting video...")
for idx, frame in enumerate(frames):
    out.write(frame)

out.release()
print("Done writing erosion video into a .mp4 file")
print(f"Total changes: {change_count}")
```

```python
# Blob Detection
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("data/images/blob_detection.jpg", cv2.IMREAD_GRAYSCALE)
imgCopy = img.copy()

detector = cv2.SimpleBlobDetector_create()

keypoints = detector.detect(img)

img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

for k in keypoints:
    x,y = k.pt
    x,y = int(round(x)), int(round(y))

    cv2.circle(img, (x,y), 5, (0,0,0), 2, cv2.LINE_AA)

    diameter = k.size
    radius = int(round(diameter/2))
    cv2.circle(img, (x,y), radius, (0,0,255), 2,cv2.LINE_AA)

    plt.imshow(img[...,::-1])
```

```python
# Setup SimpleBlobDetector parameters.
params = cv2.SimpleBlobDetector_Params()
 
# Change thresholds
params.minThreshold = 10
params.maxThreshold = 200
 
# Filter by Area.
params.filterByArea = False
params.minArea = 1500
 
# Filter by Circularity
params.filterByCircularity = True
params.minCircularity = 0.1
 
# Filter by Convexity
params.filterByConvexity = True
params.minConvexity = 0.87
 
# Filter by Inertia
params.filterByInertia = False
params.minInertiaRatio = 0.01
 
# Create a detector with the parameters
detector = cv2.SimpleBlobDetector_create(params)
```
