#### Color Histogram

```python
img = cv2.imread('data/images/Emerald_Lakes_New_Zealand.jpg')

hist1 = cv2.calcHist([img], [0], None, [256], [0, 255])
hist2 = cv2.calcHist([img], [1], None, [256], [0, 255])
hist3 = cv2.calcHist([img], [2], None, [256], [0, 255])

plt.figure(figsize=(18, 10))
plt.subplot(221);plt.imshow(img[...,::-1])
plt.subplot(222);
plt.plot(hist1, 'b');plt.plot(hist2, 'g');plt.plot(hist3, 'r')
plt.xlim([0, 256])
plt.ylim([0, 200000])
```

```python
img = cv2.imread('data/images/Emerald_Lakes_New_Zealand.jpg')

mask_hist = np.zeros((img.shape[0], img.shape[1]), dtype='uint8')

mask_hist[2100:2400, 1500:2200] = 255

mat = [mask_hist, mask_hist, mask_hist]
mask_3ch = cv2.merge(mat, 3)

img_roi = cv2.bitwise_and(img, mask_3ch)

hist1_lake = cv2.calcHist([img], [0], mask_hist, [256], [0, 255])
hist2_lake = cv2.calcHist([img], [1], mask_hist, [256], [0, 255])
hist3_lake = cv2.calcHist([img], [2], mask_hist, [256], [0, 255])

plt.figure(figsize = [18, 10])
plt.subplot(223); plt.imshow(img_roi[:, :, ::-1])
plt.subplot(224); plt.plot(hist1_lake, 'b'); plt.plot(hist2_lake, 'g'); plt.plot(hist3_lake, 'r')
plt.xlim([0, 256])
plt.ylim([0, 10000])
```

```python
img = cv2.imread('data/images/parrot.jpg', cv2.IMREAD_GRAYSCALE)

# Equalize histogram
img_eq = cv2.equalizeHist(img)

# Display the images
plt.figure(figsize = (18, 6))
plt.subplot(121); plt.imshow(img_eq, cmap='gray'); plt.title('Equalized Image')
plt.subplot(122); plt.hist(img_eq.ravel(), 256, range = [0, 256]); plt.title('Equalized Histogram')
```

#### Wrong way of histogram equalization

```python
img = cv2.imread('data/images/parrot.jpg')
img_eq = np.zeros_like(img)

# Peform histogram equalization on each channel separately.
for i in range(0, 3):
    img_eq[:, :, i] = cv2.equalizeHist(img[:, :, i])

# Display the images.
plt.figure(figsize = (18, 6))
plt.subplot(121); plt.imshow(img[:, :, ::-1]); plt.title('Original Color Image')
plt.subplot(122); plt.imshow(img_eq[:, :, ::-1]); plt.title('Equalized Image')
```

#### Histogram Equalization

```python
img = cv2.imread('data/images/parrot.jpg', cv2.IMREAD_COLOR)

# Convert to HSV.
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Perform histogram equalization only on the V channel, for value intensity.
img_hsv[:,:,2] = cv2.equalizeHist(img_hsv[:, :, 2])

# Convert back to BGR format.
img_eq = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR)

# Display the images.
plt.figure(figsize = (18, 6))
plt.subplot(121); plt.imshow(img[:, :, ::-1]); plt.title('Original Color Image')
plt.subplot(122); plt.imshow(img_eq[:, :, ::-1]); plt.title('Equalized Image')
```

```python
plt.figure(figsize = [15,4])
plt.subplot(121); plt.hist(img.ravel(),256,range = [0, 256]); plt.title('Original Image')
plt.subplot(122); plt.hist(img_eq.ravel(),256,range = [0, 256]); plt.title('Histogram Equalized')
```

#### Isolate the lake from the original image

```python
# read the image
img = cv2.imread('data/images/Emerald_Lakes_New_Zealand.jpg', cv2.IMREAD_COLOR)
plt.imshow(img[...,::-1])
```

```python
# Convert image from bgr to hsv
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Specify the range for mask
g_lb = np.array([50, 10, 10], np.uint8)
g_ub = np.array([90, 255, 255], np.uint8)

# Create the mask
g_mask = cv2.inRange(img_hsv, g_lb, g_ub)

# Plot the mask
plt.imshow(g_mask, cmap='gray')
```

```python
g_seg = cv2.bitwise_and(img, img, mask=g_mask)

plt.imshow(g_seg[...,::-1])
```
