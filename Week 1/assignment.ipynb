```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
qrCode = cv2.imread('data/images/QRCode-Satya.png', -1)
plt.imshow(qrCode[:,:,::-1])
```

```python
qrCode.shape, qrCode.dtype, type(qrCode)
```

```python
detector = cv2.QRCodeDetector()
opencvData, bbox, rectifiedImage = detector.detectAndDecode(qrCode)
bbox, bbox.shape
```

```python
for i in range(len(bbox[0])):
    pt1 = tuple(np.uint8(bbox[0][i]))
    pt2 = tuple(np.uint8(bbox[0][(i+1) % len(bbox[0])]))
    print(pt1, pt2)
    cv2.line(qrCode, pt1, pt2, color=(255,0,0), thickness=2)
```
