```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
images = []

imagefiles = ["data/images/scene/" + f for f in os.listdir("data/images/scene") if f.endswith(".jpg")]

imagefiles.sort()

destination = "scene_result.png"

plt.figure(figsize=[20,15])

for filename in imagefiles:
    img = cv2.imread(filename)
    images.append(img)

stitcher = cv2.Stitcher_create()
status, pano = stitcher.stitch(images)

if status == cv2.Stitcher_OK:
    plt.imshow(pano[...,::-1])

cv2.imwrite(destination, pano)
```
